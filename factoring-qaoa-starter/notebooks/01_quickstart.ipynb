# Factoring-derived MIS quickstart
This notebook checks the reduction and evaluates one fixed set of p=1 QAOA angles through local light cones.

```python
from factoring_qaoa.instances import build_factoring_mis
from factoring_qaoa.stats import graph_statistics
from factoring_qaoa.classical import solve_sat, decode_factors, assignment_to_independent_set
from factoring_qaoa.p1_local import p1_objective
```

```python
instance = build_factoring_mis(9, occurrence_order='clause')
graph_statistics(instance.graph)
```

```python
assignment = solve_sat(instance)
factors = decode_factors(instance, assignment)
witness = assignment_to_independent_set(instance, assignment)
factors, len(witness), instance.target_alpha
```

```python
result = p1_objective(instance.graph, gamma=0.5, beta=0.25, penalty=1.25)
result
```
